Option 1: Generate a time serie from os11 scenario input but with different astrophysical parameters

```python
from corgisim.wavefront_estimation import generate_time_series_from_os11

#Define the different astrophysical parameters here
vmag_ref = 2 #Reference star magnitude
vmag_sci = 0 #Science star magnitude
frame_exp = 1  #Single frame exposure time in sec
total_exp_time = 20  #Total exposure time in sec
x_off_mas = 300 #x offset of the point source in mas
y_off_mas = 0 #y offset of the point source in mas
dmag = 15 #Delta magnitude between star and off axis point source
sptype = 'G0V' #Spectral type of the star

#Store the information in Corgisim compatible dictionaries
ref_star_properties = {'Vmag': vmag_ref, 'spectral_type': sptype, 'magtype': 'vegamag'}
target_star_properties = {'Vmag': vmag_sci, 'spectral_type': sptype, 'magtype': 'vegamag'}
point_source_info = [{
    'Vmag': vmag_sci + dmag,
    'magtype': 'vegamag',
    'position_x': x_off_mas,
    'position_y': y_off_mas
}]

#Change the path to save the PSF at the desired location
outdir = None #"path/to/save/outputs"

#Call the main function to generate simulation from os11 wavefront errors series
#Can change the coronagraph, bandpass, polaxis and dark hole performance here
generate_time_series_from_os11(ref_star_properties, target_star_properties, point_source_info, frame_exp,
                               total_exp_time, outdir, cor_type='hlc', bandpass='1F', polaxis=10,
                               dm_case='3e-8')
```
